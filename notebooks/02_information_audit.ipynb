# Phase 2: Information Structure Analysis

Before building any ensemble, the most important question is whether your candidate inputs are actually measuring different things. Redundant inputs don't improve predictions — they just add noise and complexity. This notebook tests each candidate source for independent predictive signal using pairwise correlation analysis, incremental predictive value testing, and category-specific accuracy profiling. The output is a principled, justified decision about which sources enter the ensemble.

## Oscar Context

This notebook evaluates the candidate Oscar prediction sources — precursor ceremonies, Gold Derby expert consensus, and prediction markets — to determine which carry genuinely independent signal and which are redundant. The goal is to enter only non-redundant, informative sources into the ensemble built in Phase 4.
