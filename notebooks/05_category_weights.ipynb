# Phase 5: Global vs. Category-Specific Weights

A recurring question in applied forecasting is whether a single global model outperforms segment-specific models. Fitting separate models per segment allows them to capture local patterns, but at the cost of increased complexity and higher variance estimates. This notebook tests that tradeoff in the Oscar context: do sources carry different information value for different category types, or is a single universal weight set sufficient?

## Oscar Context

Oscar categories can be grouped by type — acting, crafts, above-the-line, etc. — and it is plausible that prediction markets outperform precursor ceremonies for some groups while the reverse holds for others.
