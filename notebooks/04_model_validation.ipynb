# Phase 4: Ensemble Weight Optimization

When multiple imperfect forecasting sources are available, combining them into a weighted ensemble can outperform any individual source — but only if the weights are set correctly. Intuition-based blending is a common starting point; this notebook tests whether systematic optimization produces meaningfully better results. Four models are compared under leave-one-year-out cross-validation: naive equal weighting, manually tuned weights, the single best source, and a constrained-optimization ensemble.

## Oscar Context

This notebook builds and validates ensemble models for Oscar prediction. Using the retained sources from Phase 2 and the calibration findings from Phase 3, we optimize a weighted blend of source probabilities to minimize Brier Score, then compare it against simpler baselines using leave-one-year-out cross-validation.
