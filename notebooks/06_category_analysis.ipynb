# Phase 6: Category Predictability Analysis

Not all forecasting problems are equally hard. Understanding where a model is reliable and where it isn't — and identifying the features that drive that variation — is a critical analytical skill. A forecast system that surfaces its own uncertainty is more useful than one that treats all predictions equally. This notebook characterizes which prediction targets are inherently difficult, tests hypotheses about what drives difficulty, and attempts to cluster targets by predictability.

## Oscar Context

Some Oscar categories are perennially predictable (Best Picture frontrunners often telegraph months in advance) while others are famously volatile (certain craft categories can swing late). This notebook ranks Oscar categories by historical predictability, tests what structural features explain the variance in difficulty, and clusters categories to identify groups that share predictability profiles.
